# DSC 495: Final Project Data Cleaning

```python
import pandas as pd
import os
```

## Reading in Data

```python
directory = './data/raw_data/'
files = os.listdir(directory)
datasets = []

for file in files: 
    datasets.append(pd.read_csv(directory + file))
```

## Data Cleaning

```python
#combine datasets
flight_delays = pd.concat(datasets)
flight_delays.head()
```

```python
#create variable for total delay in minutes and name of month
month_dictionary = {1: 'January', 
                    2: 'February', 
                    3: 'March', 
                    4: 'April',
                    5: 'May', 
                    6: 'June', 
                    7: 'July', 
                    8: 'August', 
                    9: 'September', 
                    10: 'October', 
                    11: 'November', 
                    12: 'December'}
        
flight_delays['MONTH_NAME'] = flight_delays['MONTH'].apply(lambda x: month_dictionary[x])   
flight_delays['TOT_DELAY'] = flight_delays['CARRIER_DELAY'] + flight_delays['WEATHER_DELAY'] + flight_delays['NAS_DELAY'] + flight_delays['SECURITY_DELAY'] + flight_delays['LATE_AIRCRAFT_DELAY']

flight_delays
```

```python
flight_delays.to_csv('./data/processed_data/flight_delays.csv')
```
